```python
import pandas as pd
from datetime import datetime
import plotly.express as px

file_path = r"C:\Users\satya\Downloads\Play Store Data.csv"
df = pd.read_csv(file_path)

df['Last Updated'] = pd.to_datetime(df['Last Updated'], errors='coerce')

def convert_installs(installs_str):
    if isinstance(installs_str, str):
        installs_str = installs_str.replace(',', '')  
        if installs_str.lower() == 'free':  
            return 0
        if 'M' in installs_str:
            return float(installs_str.replace('M', '')) * 1_000_000
        elif 'K' in installs_str:
            return float(installs_str.replace('K', '')) * 1_000
        else:
            try:
                return float(installs_str.replace('+', ''))  
            except ValueError:
                return None 
    return None

df['Installs'] = df['Installs'].apply(convert_installs)
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')

start_date = datetime(2018, 1, 1)  
end_date = datetime(2019, 12, 31)  

filtered_df = df[
    (df['Last Updated'] >= start_date) & 
    (df['Last Updated'] <= end_date) &    
    (df['Installs'] >= 100000) &        
    (df['Reviews'] > 1000) &          
    (~df['Genres'].str.startswith(('A', 'F', 'E', 'G', 'I', 'K')))  
]

filtered_df_cleaned = filtered_df.dropna(subset=['Installs', 'Rating', 'Reviews'])
correlation_matrix = filtered_df_cleaned[['Installs', 'Rating', 'Reviews']].corr()

current_time = datetime.now().time()

start_time = datetime.strptime('15:00', '%H:%M').time()  
end_time = datetime.strptime('18:00', '%H:%M').time()    

html_output_path = r"C:\Users\satya\OneDrive\Desktop\Satyam\Task 3.html"

insights = """
The correlation coefficient between Installs and Reviews is approximately 0.605.
This indicates a moderate positive correlation, meaning that apps with a higher number of installs tend to receive more reviews.
"""

if start_time <= current_time <= end_time:
    fig3 = px.imshow(correlation_matrix, text_auto=True, color_continuous_scale='RdBu_r',
                     title="Correlation Matrix Heatmap: Installs, Ratings, and Reviews")
    fig3.write_html(html_output_path,insights)
    fig3.show()  
    print(f"Heatmap and Insights saved to: {html_output_path}")
    print(f"Insight: {insights}")
else:
    with open(html_output_path, 'w') as f:
        f.write("<h1>Come back between 3 PM to 6 PM</h1>")
    print("Heatmap not displayed. Please come back between 3 PM to 6 PM.")
    print(f"Message saved to: {html_output_path}")
```
